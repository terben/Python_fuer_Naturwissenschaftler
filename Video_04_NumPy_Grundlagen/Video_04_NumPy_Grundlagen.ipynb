### 2. Grundideen von NumPy-Arrays

NumPy stellt primär einen Python Container, das `NumPy-Array` bereit. Es ist ein homogener (primär Zahlen) Container, der **sehr schnelle**, simultane Vektoroperationen auf ganzen Arrays **ohne** Schleifenkonstrukte zur Verfügung stellt.  

```python
# Numerische Operationen auf Python-Listen müssen mit expliziten Schleifen
# realisiert werden: 
li = [1, 2, 3, 4]

for n in range(len(li)):
    li[n] = li[n]**2

li
```

```python
# NumPy Arrays erlauben *elementweise* mathematische Operationen **ohne** Schleifen:
x = np.array([1, 2, 3, 4])
x = x**2

x
```

Vergleich der Performance zwischen Python-Listen und NumPy-Arrays: Operationen auf Datenstrukturen mit einer Million Elemente:

```python
# Große Listen mit einer Million Elementen:
size = 1_000_000

python_list = list(range(size))
numpy_array = np.arange(size)
```

```python
%%timeit

python_list = list(range(size))

# Python list (for-Schleife)
for i in range(size):
    python_list[i] = python_list[i] + 1
```

```python
%%timeit

numpy_array = np.arange(size)

# NumPy array (Vektoroperation)
result_numpy = numpy_array + 1
```

**Elementweise** Operationen funktionieren auch mit allen gängigen mathematischen Funktionen und auch zwischen Arrays:

```python
# Operationen auf mathematischen Funktionen:
x = np.array([1, 2, 3, 4])

1 / x + np.exp(x) * np.sin(x) + 5
```

```python
# Operationen zwischen Arrays:
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

(a + b) / a
```

NumPy stellt neben der Arrayfunktionalität **sehr** viele Funktionen für die Arbeit mit Arrays bereit. Die wichtigsten werden wir im Laufe dieser Videoserie kennenlernen:

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Mittelwert eines Arrays
print(np.mean(a))

# Skalarprodukt zweier Arrays
print(np.sum(a * b)) # Berechnung per Hand
print(np.dot(a, b))  # NumPy-Funktion 
```

### 3. Erzeugung von NumPy-Arrays und einfache Plots

Hier einige Beispiele zur Erzeugung von NumPy-Arrays:

```python
# Methode 1: Aus Python Liste:
arr1 = np.array([1, 2, 3, 4, 5])
print("Aus Liste:", arr1)

# Methode 2: Arrays aus Einsen
ones = np.ones(5)
print("Ones:", ones)

# Methode 3: Arrays aus Zufallszahlen
random_arr = np.random.rand(5)  # Gleichförmige Verteilung [0, 1)
print("Zufällig (Gleichverteilung):", random_arr)

random_int = np.random.randint(1, 10, size=5)  # [low, high), size
print("Zufällig (Ganzzahlen):", random_int)
```

Anfangs sind `np.linspace` und `np.arange` für Sie unter den wichtigsten Methoden zur Arrayerzeugung:

```python
# linspace: Anzahl der Elemente zwischen Anfangs- und Endpunkt:
x = np.linspace(0., 1., 21)
x
```

```python

# arange: Explizite Angabe des Abstands der Intervallelemente
# (Endwert ist nicht(!) enthalten)
x = np.arange(0., 1., 0.05)
x
```

```python
# Berechnung einer Funktion an allen Arraypunkten
y = np.sin(np.pi * x)
y
```

Und ein einfacher Plot

```python
# Plots von NumPy-Arrays bestehen aus einzelnen Punkten!
plt.plot(x, y)
```

```python
# Plots, die man für Dokumente verwenden möchte, sollten kommentiert werden
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('x vs y')
plt.grid()
```

### 5. Boolsche Masken / Boolsche Indizierung

Unter den nützlichsten Features von NumPy Arrays sind boolsche Masken. Sie erlauben eine weit größere Flexibilität bei der Auswahl von Array Elementen als Slicing. Wir können unter anderem alle Elemente auswählen, die bestimmte mathematische Bedingungen erfüllen:

```python
x = np.linspace(1, 4, 20)
y = np.sin(x)

plt.plot(x, y)
```

Im folgenden wollen wir nur mit Elementen arbeiten, die $x>2$ erfüllen:

```python
x
```

```python
x > 2
```

Diese *boolsche Maske* kann zur Indizierung von $x$ und/oder $y$ verwendet werden

```python
mask = (x > 2)
x, x[mask]
```

Als Plot

```python
plt.plot(x, y)
plt.plot(x[mask], y[mask])
```

Boolsche Masken können kombiniert werden

```python
# Unter mathematischen Operationen gilt 'False = 0' und 'True = 1'
mask = (x > 2) & (x < 3)  # (x > 2) *und* (x < 3)
plt.plot(x, y)
plt.plot(x[mask], y[mask])
```

```python
mask = (x < 1.5) | (x > 3)  # (x < 1.5) *oder* (x > 3)
plt.plot(x, y)
plt.plot(x[mask], y[mask], '.')
```

### 6. Mehrdimensionale NumPy-Arrays

Python‑Listen sind immer *eindimensionale* Container, können aber geschachtelte Listen enthalten. NumPy hingegen stellt echte *mehrdimensionale* Arrays bereit.

```python
li = [[1, 2, 3, 4], [11, 22, 33, 44], [1, 4, 9, 16]]
li
```

```python
x = np.array([[1, 2, 3, 4], [11, 22, 33, 44], [1, 4, 9, 16]])
x
```

Am besten besorgen Sie sich von einem unbekannten Array erst einmal die Dimensionen

```python
x.shape
```

Dies besagt, dass das Array $x$ zweidimensional ist. Die erste Dimension hat 3 Elemente (3 Zeilen) und die zweite Dimension hat 4 Elemente (4 Spalten). 

**Anmerkung:** Ein NumPy-Array besteht neben den Daten noch aus Metadaten (Dimension, Datentyp, ...). Die allerwichtigsten für den Anfang sind:

```python
# Arraydimension
print(f"Arraydimension: {x.ndim}")

# Arrayshape
print(f"Shape: {x.shape}")

# Datentyp
print(f"Datentyp: {x.dtype}")

# Arrayelemente
print(f"Elemente im Array: {x.size}")

# Elementgröße (Bytes)
print(f"Elementgröße: {x.itemsize} byte")

print(f"Arraygröße: {x.nbytes} byte")
```

* Die Indizierung für den Elementzugriff erfolgt jetzt auch mehrdimensional

```python
x
```

```python
x[0, 2]   # Element der *ersten* Zeile und der dritten Spalte
```

* Indiziere die erste Spalte (alle Werte der ersten Dimension und den ersten Wert der zweiten Dimension)

```python
x[:,0]
```

* Indiziere eine Teilmatrix

```python
x[1:3, 1:3]  # Slices sind immer 'ohne' das letzte Element!
```

Oft ist es nützlich, ein zweidimensionales Array zu transponieren (Austausch von Zeilen und Spalten)

```python
x, x.T # x.T steht fuer die Transponierte des Arrays x
```

```python
x.shape, x.T.shape
```

```python
x[0,:]
```

NumPy und andere Module bieten Routinen, um wissenschaftliche Daten *aller* gängiger Formate in Arrays einzulesen und auch wieder abzuspeichern (Module).
